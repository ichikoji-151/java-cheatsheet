# Bacic Structure

- ノートブック上の環境で下のソースコードはクラスを定義しただけ
- 表示するだけなら2個めのソースコードのみでも実行できる
- mainを呼びたす際は下のソースコード
- 今後は上に実際に使うソースコード、下には実行するための処理を記述する

 - System : 標準入出力、標準エラー、プロパティ、環境変数へのアクセス、ファイルやライブラリのローディングや配列やコピーなどの機能を持つ`クラス`
 - out : `System`クラスで定義されている`PrintStream`型の`static``フィールド`
 - println : 引数の値を標準出力の新しい一行へ出力する`メソッド` 

```java
public class HelloWorld {
    public static void main(String[] args) {
        System.out.println("Hello World");
    }
}
```

```java
System.out.println("Hello World");
```

```java
HelloWorld.main(new String[0]);
```

## エスケープシーケンス
特別な記号や出力方法を制御するために記号(\)を使う
|記号の組み合わせ|意味|
|---|---|
|`\'`|'|
|`\"`|"|
|`\\`|\|
|`\n`|改行(ラインフィードLF)|
|`\t`|水平タブ|
|`\r`|復帰(キャリッジ・リターン：CR)|

## 変数

```java
int i; // 変数の宣言
i = 5; // 値の代入
System.out.println(i); // 値の参照
```

## 算術演算子と優先順位
剰余から順に優先される
|演算子|演算の内容|
|---|---|
|+|加算|
|-|減算|
|*|乗算|
|/|除算|
|%|剰余|

## 算術演算の短縮表現に用いる演算子
|演算子|演算の内容|
|---|---|
|+=|加算代入|
|-=|減算代入|
|*=|乗算代入|
|/=|乗算代入|
|%=|剰余代入|
|++|インクリメント|
|--|デクリメント|

## Javaで使用される型
|型|格納できる値の範囲|
|---|---|
|char|1文字(16ビット、Unicode文字)|
|boolean|真偽値。trueまたはfalseのどちらか|
|byte|8ビット符号付き整数|
|short|16ビット符号付き整数|
|int|32ビット符号付き整数|
|long|64ビット符号付き整数|
|float|32ビット符号付き浮動小数点数|
|double|64ビット符号付き浮動小数点数|

double > float > long > int > short > byte
